## Question 11 – Operator Precedence

Predict and then print the answers to `5 + 2 * 3` and `(5 + 2) * 3`. Add a comment explaining why the answers are different.

```python
# Question 11 – Operator Precedence
print("Result of 5 + 2 * 3:", 5 + 2 * 3)      # Predicted: 11
print("Result of (5 + 2) * 3:", (5 + 2) * 3)  # Predicted: 21

# The answers are different because of operator precedence:
# Multiplication (*) has higher precedence than addition (+), so in the first expression 2 * 3 is calculated first.
# Parentheses () have the highest precedence, forcing 5 + 2 to be calculated first in the second expression.
```

## Question 13 – Short-Circuit Evaluation

Print the results of `False and True` and `True or False`. In a comment, explain briefly why Python may not need to evaluate the second part of an `and` or `or` expression.

```python
# Question 13 – Short-Circuit Evaluation
print("False and True:", False and True)
print("True or False:", True or False)

# Short-circuit evaluation:
# For 'and', if the first operand is False, the overall expression is guaranteed to be False, so Python skips evaluating the second part.
# For 'or', if the first operand is True, the overall expression is guaranteed to be True, so Python skips evaluating the second part.
```

## Question 14 – Simple if-else Statement

Ask the user to enter their age. If the age is 18 or above, print `Eligible to vote`; otherwise, print `Not eligible to vote`.

```python
# Question 14 – Simple if-else Statement
try:
    user_age = int(input("Enter your age: "))
    if user_age >= 18:
        print("Eligible to vote")
    else:
        print("Not eligible to vote")
except ValueError:
    print("Please enter a valid integer for age.")
```

## Question 15 – if-elif-else and Nested if

Ask the user to enter marks out of 100. Use `if-elif-else` to print `Excellent` for 80 or above, `Good` for 60–79, `Pass` for 40–59, and `Fail` below 40. First check that marks are between 0 and 100. For valid marks, use a nested `if` to display `Passed the examination` when marks are at least 40.

```python
# Question 15 – if-elif-else and Nested if
try:
    marks = float(input("Enter your marks out of 100: "))

    # Outer check for valid marks range
    if 0 <= marks <= 100:
        # Check grade status
        if marks >= 80:
            print("Excellent")
        elif marks >= 60:
            print("Good")
        elif marks >= 40:
            print("Pass")
        else:
            print("Fail")

        # Nested check for passing
        if marks >= 40:
            print("Passed the examination")
    else:
        print("Invalid marks! Please enter a value between 0 and 100.")
except ValueError:
    print("Please enter a valid numeric value.")
```
