The 7 rules of Python identifiers
Rule	Explanation	Valid example	Invalid example
1. Start with a letter or underscore	The first character must be an alphabet or _.	name, _age	2name
2. Cannot start with a number	Numbers are allowed after the first character.	age2	2age
3. Only letters, numbers, and underscores	Special characters are not allowed.	student_1	student-name
4. No spaces allowed	Use underscores instead of spaces.	first_name	first name
5. Case-sensitive	Uppercase and lowercase letters are different.	age, Age, AGE	These are not the same identifier.
6. Cannot use Python keywords	Reserved words cannot be used as identifiers.	marks, total	if, class, for
7. Can be of any reasonable length	Python does not impose a small fixed length limit on identifiers.	student_total_marks	No specific length-based invalid example.
